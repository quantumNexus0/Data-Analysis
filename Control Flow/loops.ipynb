# Looping in python
Looping means repeating a block of code multiple times. 
Python mainly has two types of loops:
1. **for loop**
2. **while loop**


# Loop Control Statements
Python mainly has 3 loop control statements:
| Statement  | Meaning                |
| ---------- | ---------------------- |
| `break`    | 🛑 Stop the loop       |
| `continue` | ⏭️ Skip this iteration |
| `pass`     | ➖ Do nothing           |


```python
## for loop 
# range(start, stop) 
# range(start, stop, step)

for i in range(0, 5):
    print(i)
print("-----------------------------------")
for i in range(2, 10, 2):  #Here, 2 is the starting number, 10 is the stopping point, and 2 is the step
    print(i)
   
```

```python
# printing a revers number
for i in range(10,0,-1):
    print(i)
```

```python
## strings

Str="hello world"
for chr in Str:
    print(chr)
```

```python
# While loop
# the while loop continues to execute as long as the condition is true.

count=0
while count<10:
    print(count)
    count=count+1
```

```python
count = 0
while count%2==0:
    print(count)
    count = count+1
```

```python
## loop control statements
## break : It completely terminates the loop.
for i in range(1, 10):
    if i == 5:
        break
    print(i)
print("-----------------------------------------------")
## continue : It skips the current iteration and moves to the next one.
for i in range(1, 6):
    if i == 3:  #When i == 3, print(i) is skipped.
        continue
    print(i)

print("-----------------------------------------------")
## pass — Do nothing
for i in range(5):
    if i == 3:
        pass
    print(i)
```

```python
## check number is even or not 
for i in range(10):
    if i%2==0:
        continue  #Agar number even hai, toh is iteration ko skip karo aur next number par chale jao.
    print(i)
```

```python
## Nested loops 
## a loop inside a loop

for i in range(3):
    for j in range(2):
        print(f"i:{i} and j:{j}")
```

```python

# Using while loop

# n = int(input("Enter a number:"))
n = 10

sum = 0
count = 1

while count <= n:
    sum = sum + count
    count = count + 1

print(f"Sum of first {n} natural numbers is: {sum}")
```
