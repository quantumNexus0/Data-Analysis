```python
## understanding variable types
# Using the type() function to check variable types
# Python is dynamically typed, type of a variable is determined at runtime based on the value assigned to it.
name = "Alice"      # String
age = 25            # Integer
height = 5.7        # Float
is_student = True   # Boolean

print(type(name))
print(type(age))
print(type(height))
print(type(is_student))
```

### Type checking and type conversion (type casting)
Type conversion means changing a value from one data type to another.

Type conversion : 
 1. Implicit Type Conversion 
 2. Explicit Type Conversion 
 
| Feature               | Implicit Type Conversion                   | Explicit Type Conversion                       |
| --------------------- | ------------------------------------------ | ---------------------------------------------- |
| **Meaning**           | Python automatically changes the data type | Programmer manually changes the data type      |
| **Done by**           | Python interpreter                         | Programmer                                     |
| **Need function?**    | No                                         | Yes, usually `int()`, `float()`, `str()`, etc. |
| **Risk of data loss** | Generally low                              | Data loss can occur                            |
| **Example**           | `10 + 2.5 → 12.5`                          | `int("10") → 10`                               |



### In Explicit Type conversion Programmer manually changes the data type and Python provides built-in functions for this.
### Common Type Conversion Functions 

| Function  | Converts to | Example                         |
| --------- | ----------- | ------------------------------- |
| `int()`   | Integer     | `int("10")` → `10`              |
| `float()` | Float       | `float("10.5")` → `10.5`        |
| `str()`   | String      | `str(100)` → `"100"`            |
| `bool()`  | Boolean     | `bool(1)` → `True`              |
| `list()`  | List        | `list("ABC")` → `['A','B','C']` |
| `tuple()` | Tuple       | `tuple([1,2,3])` → `(1,2,3)`    |


```python
### Example 1: String → Integer
# This is an example of implicit type conversion in Python.

age = "25"  # String
print(age, type(age))  # Output: 25 <class 'str'>

age = 34
print(age, type(age))  # Output: 34 <class 'int'>

name = "Alice"
print(name, type(name))  # Output: Alice <class 'str'>


# This is an example of explicit type conversion in Python.
# Converting String to Integer
age = int(age)  # Convert to Integer
print(age, type(age))  # Output: 25 <class 'int'>

age = float(age)
print(age, type(age))  # Output: 25.0 <class 'float'>


## Lossy Conversion in Python = Type conversion in which some information/data is lost.

## Lossy conversion means converting a value from one data type to another where some information is lost during the conversion.

#Example: Float → Integer
x = 10.75
y = int(x)
print(y)  # Output: 10 Here, 0.75 is lost:

##10.75  →  10  → decimal part lost
```

```python
### Type checking
# Type checking means finding out what data type a variable contains.
# type() function is used to check the type of a variable in Python.
# isinstance() function is used to check if a variable is an instance of a specific class or data type. and It checks whether a value belongs to a particular data type.
name = "Rahul"
age = 21
marks = 85.5
student = True

print(type(name))
print(type(age))
print(type(marks))
print(type(student))

# Using isinstance() function
print(isinstance(name, str))      # Output: True

age = 21
print(isinstance(age, int))     # Output: True

age = 21
print(isinstance(age, str))   # Output: False
```
