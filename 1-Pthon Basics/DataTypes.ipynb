# Main Data  types in python

| Data Type  | Keyword    | Example             |
| ---------- | ---------- | ------------------- |
| Integer    | `int`      | `10`, `-5`, `100`   |
| Float      | `float`    | `10.5`, `3.14`      |
| Complex    | `complex`  | `2 + 3j`            |
| String     | `str`      | `"Hello"`           |
| Boolean    | `bool`     | `True`, `False`     |
| List       | `list`     | `[1, 2, 3]`         |
| Tuple      | `tuple`    | `(1, 2, 3)`         |
| Set        | `set`      | `{1, 2, 3}`         |
| Dictionary | `dict`     | `{"name": "Rahul"}` |
| None       | `NoneType` | `None`              |


```python
# Easy Classification
 # numbers ----> int, float, complex
 # strings ----> str
 # boolean ----> bool
 # collections ----> list, tuple, set, dict
 # specialized ----> NoneType

#numbers examples
age = 21
number = -10j
x = 2 + 3j
print(type(x) , x)
marks = 99.9
print(type(age))
print(type(number))
print(type(marks))
print("-------------------------------")
#Strings examples
name = "John"
color = 'Red'
print(type(name))
print(type(color))
print("-------------------------------")

#Boolean examples
is_student = True
is_working = False
print(type(is_student))
print(type(is_working))
print("-------------------------------")

#Collections examples
#List
fruits = ["Apple", "Mango", "Banana"]
print(fruits)
print(type(fruits))

#Tuple
fruits_tuple = ("Apple", "Mango", "Banana")
print(fruits_tuple)
print(type(fruits_tuple))

#Set
fruits_set = {"Apple", "Mango", "Banana"}
print(fruits_set)
print(type(fruits_set))

#Dictionary
fruits_dict = {"Apple": 5, "Mango": 10, "Banana": 15}
print(fruits_dict)
print(type(fruits_dict))
```
