#### Syntax → Structure
#### Semantics → Meaning

### Syntax (How the program is written) 
 Syntax refers to the rules and structure used to write a valid program.
Example in C:

int x = 10;

This is syntactically correct.

But:

int x = 10

is syntactically incorrect because the semicolon ; is missing.  

### Semantics(What the program means/does)
Semantics refers to the meaning or behavior of a program.

A program can have correct syntax but still have incorrect or unintended meaning.

Example:

int x = 10;
int y = 0;
int z = x / y;

The syntax is correct, but the statement x / y causes a division-by-zero error.

### Difference Between Syntax and Semantics
| Syntax                                         | Semantics                                                 |
| ---------------------------------------------- | --------------------------------------------------------- |
| Deals with structure/grammar                   | Deals with meaning                                        |
| Determines whether code is valid               | Determines what valid code does                           |
| Syntax errors are usually detected by compiler | Semantic errors may occur during compilation or execution |
| Example: missing `;`                           | Example: division by zero                                 |
| **How** the program is written                 | **What** the program does                                 |

### Indentation 
Indentation in python is used to define the structure and hierarchy of the code. Unlike many other programming languages that use braces{} to delimit blocks of code, python uses indentation to determine the grouping of statements. this means that all the statements within a block must be indented at the same level.

```python
##Indentation- python uses indentation to define blocks of code. Indentation is typically done using spaces or tabs, and it is important to be consistent with the indentation style used in your code.

age=32
if age>18:
    print("You are an adult")

print("This line is not indented and will always execute")
```

```python
## Line Continuation- python allows you to break long lines of code into multiple lines for better readability. You can use the backslash (\) character to indicate that a line should continue onto the next line.

total = 10 + 20 + 30 + \
        40 + 50
print(total)
```

```python
## multiple statements on a single line- python allows you to write multiple statements on a single line by separating them with semicolons (;). However, it is generally recommended to write each statement on a separate line for better readability.
x=5; y=10; z=x+y
print(z)
```
